## 3. Quick look at the data (2-3 real numbers)

I will examine the starter dataset to check whether there are enough pages with meaningful search exposure to make CTR opportunity scoring worth investigating. I will report the number of eligible pages and compare their mean CTR across position tiers. These measurements will provide an initial indication of the available data and whether position-adjusted comparisons are worth exploring. I will use the observed results rather than assume the relationship in advance.

```python
import pandas as pd

url = (
    "https://raw.githubusercontent.com/"
    "flyrank-bih/flyrank-ml-internship-starter/"
    "main/data/raw/content_refresh_anonymized.csv"
)

df = pd.read_csv(url)

# Keep pages with meaningful impressions and a valid position.
eligible = df[
    (df["impressions_90d"] >= 100) &
    (df["avg_position"] > 0)
].copy()

print("Total pages in starter dataset:", len(df))
print("Pages with >=100 impressions and valid position:", len(eligible))
print("\nMean CTR and page count by position tier:")

summary = (
    eligible.groupby("position_tier")
    .agg(
        page_count=("ctr", "size"),
        mean_ctr=("ctr", "mean")
    )
    .sort_values("mean_ctr", ascending=False)
)

print(summary.round(3).to_string())
```
